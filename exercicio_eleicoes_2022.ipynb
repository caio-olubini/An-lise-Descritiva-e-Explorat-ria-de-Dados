## 1 · Os dados

Os arquivos do TSE vêm em `.zip`, um por estado, e dentro de cada um há um
arquivo `BRASIL` que junta todos. A célula abaixo baixa os dois arquivos
nacionais, abre o zip **na memória** — nada é salvo no seu computador — e
monta a tabela `cand`.

O arquivo de bens tem uma linha por **bem declarado**, não por candidato:
quem tem casa, carro e conta no banco aparece três vezes. A célula soma
isso por candidato e traz duas colunas novas para a tabela: `n_bens` e
`patrimonio`.


```python
# Instala o que faltar e monta a tabela do exercício (Colab e Jupyter).
# Rode esta célula uma vez no começo — ela baixa cerca de 10 MB do TSE.
%pip install -q pandas seaborn matplotlib

import io
import urllib.request
import zipfile

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

URL_CANDIDATOS = "https://cdn.tse.jus.br/estatistica/sead/odsele/consulta_cand/consulta_cand_2022.zip"
URL_BENS = "https://cdn.tse.jus.br/estatistica/sead/odsele/bem_candidato/bem_candidato_2022.zip"

ARQUIVO_CANDIDATOS = "consulta_cand_2022_BRASIL.csv"
ARQUIVO_BENS = "bem_candidato_2022_BRASIL.csv"

pacote = zipfile.ZipFile(io.BytesIO(urllib.request.urlopen(URL_CANDIDATOS, timeout=180).read()))
cand = pd.read_csv(pacote.open(ARQUIVO_CANDIDATOS), sep=";", encoding="latin-1", low_memory=False)

pacote_bens = zipfile.ZipFile(io.BytesIO(urllib.request.urlopen(URL_BENS, timeout=180).read()))
bens = pd.read_csv(pacote_bens.open(ARQUIVO_BENS), sep=";", encoding="latin-1", decimal=",")

# Uma linha do arquivo de bens é um bem. Somando por candidato, viram duas colunas.
patrimonio = bens.groupby("SQ_CANDIDATO")["VR_BEM_CANDIDATO"].sum()
patrimonio.name = "patrimonio"

n_bens = bens.groupby("SQ_CANDIDATO")["VR_BEM_CANDIDATO"].size()
n_bens.name = "n_bens"

cand = cand.merge(patrimonio, on="SQ_CANDIDATO", how="left")
cand = cand.merge(n_bens, on="SQ_CANDIDATO", how="left")

cand["idade"] = 2022 - pd.to_datetime(cand["DT_NASCIMENTO"], format="%d/%m/%Y", errors="coerce").dt.year

colunas = [
    "SQ_CANDIDATO", "NM_CANDIDATO", "DS_CARGO", "SG_UF", "NM_UE", "SG_PARTIDO",
    "DS_GENERO", "DS_COR_RACA", "DS_GRAU_INSTRUCAO", "DS_ESTADO_CIVIL", "DS_OCUPACAO",
    "NR_TURNO", "DS_SITUACAO_CANDIDATURA", "DS_SIT_TOT_TURNO",
    "idade", "n_bens", "patrimonio",
]
cand = cand[colunas]

print("Carregado.")
print(cand.shape)
```

### O dicionário das colunas

| coluna | o que é |
|---|---|
| `SQ_CANDIDATO` | número que identifica o candidato |
| `NM_CANDIDATO` | nome completo |
| `DS_CARGO` | cargo disputado — 10 valores, de `PRESIDENTE` a `DEPUTADO DISTRITAL` |
| `SG_UF`, `NM_UE` | unidade da federação; para presidente, `BR` / `BRASIL` |
| `SG_PARTIDO` | sigla do partido — 32 em 2022 |
| `DS_GENERO` | `MASCULINO`, `FEMININO` ou `NÃO DIVULGÁVEL` |
| `DS_COR_RACA` | cor/raça declarada |
| `DS_GRAU_INSTRUCAO` | escolaridade declarada — 8 níveis |
| `DS_ESTADO_CIVIL` | estado civil declarado |
| `DS_OCUPACAO` | ocupação declarada — 218 valores |
| `NR_TURNO` | 1 ou 2 |
| `DS_SITUACAO_CANDIDATURA` | `APTO` ou `INAPTO` |
| `DS_SIT_TOT_TURNO` | resultado da candidatura — 7 valores possíveis |
| `idade` | em anos, calculada a partir da data de nascimento |
| `n_bens` | quantos bens o candidato declarou |
| `patrimonio` | soma dos bens declarados, em reais |

`n_bens` e `patrimonio` vêm do arquivo de bens. Quem não declarou bem
nenhum fica **vazio** nas duas colunas — vazio não é zero, é ausência de
declaração.
